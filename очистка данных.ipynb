```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_excel('Online Retail.xlsx')
df
```

```python
df.info()
```

```python
df_clear = df.drop_duplicates()
df_clear
```

```python
df_clear = df_clear[df_clear['UnitPrice'] > 0]
df_clear = df_clear.dropna(subset = ['CustomerID'])
df_clear = df_clear[df_clear['Quantity'] > 0]
df_clear = df_clear.dropna(subset = ['Description'])
df_clear = df_clear[~df_clear['InvoiceNo'].astype(str).str.startswith('C')]
df_clear
```

```python
df_clear.info()
```

```python
df_clear['Revenue'] = df_clear['Quantity'] * df_clear['UnitPrice']
df_clear
```

```python
df_clear.to_csv('Очищенные данные.csv', index=False)
df_clear.to_pickle('data_clean.pkl')
```
