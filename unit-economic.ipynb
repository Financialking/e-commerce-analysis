```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_pickle('RFM анализ таблица.pkl')
data
```

```python
ltv_base = data.groupby('Segment')['Monetary'].mean().sort_values(ascending = False)
ltv_base
```

```python
churn_rate = 0.2
lifetime = 1/churn_rate

ltv_predict = lifetime*ltv_base
print(ltv_predict)
```

```python
r = 0.15
lifetime = 5
ltv_discounted = ltv_predict*0
for i in range(lifetime):
    ltv_discounted = ltv_discounted + ltv_base/(1+r)**i
print(ltv_discounted)
```

```python
ltv_all = pd.DataFrame({
    'Исторический ltv':ltv_base,
    'Прогнозный ltv': ltv_predict,
    'Дисконтированный ltv': ltv_discounted
})
print(ltv_all)
```

```python
CAC = 2000 #предположительный
ltv_all['ltv_cac_ratio'] = ltv_all['Дисконтированный ltv']/CAC
print(ltv_all[['Дисконтированный ltv', 'ltv_cac_ratio']])
```
