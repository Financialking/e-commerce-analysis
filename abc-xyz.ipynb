```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_pickle('data_clean.pkl')
data
```

```python
data = data[data['StockCode'] != 'POST']
data = data[data['StockCode'] != 'DOT']
data = data[data['StockCode'] != 'M']
data = data[data['StockCode'] != 'BANK CHARGES']
data
```

```python
ABC = data.groupby('StockCode').agg(
    Revenue_stock = ('Revenue', 'sum')
)
ABC = ABC.sort_values(by='Revenue_stock', ascending= False)
ABC
```

```python
ABC['Piece'] = ABC['Revenue_stock']/ABC['Revenue_stock'].sum()
ABC['Cum_share'] = ABC['Piece'].cumsum()
ABC
```

```python
def abc_analys(abc_class):
    if abc_class <=0.8:
        return 'A'
    elif abc_class >0.8 and abc_class <=0.95:
        return 'B'
    else:
        return 'C'
ABC['ABC_class'] = ABC['Cum_share'].apply(abc_analys)
ABC
```

```python
ABC['ABC_class'].value_counts()
```

```python
data['data_mounth'] = data['InvoiceDate'].dt.to_period('M')
data
```

```python
XYZ = data.groupby(['StockCode','data_mounth']).agg(
    Sales = ('Quantity', 'sum')
)
XYZ
```

```python
XYZ = XYZ.groupby('StockCode').agg(
    mean_sales = ('Sales','mean'),
    std_sales = ('Sales','std')
)
XYZ
```

```python
XYZ['cv'] = XYZ['std_sales']/XYZ['mean_sales']
XYZ
```

```python
def classification_xyz(xyz_class):
    if xyz_class < 0.5:
        return 'X'
    elif xyz_class >= 0.5 and xyz_class < 1:
        return 'Y'
    elif xyz_class == 'NaN':
        return 'Z'
    else:
        return 'Z'
XYZ['XYZ'] = XYZ['cv'].apply(classification_xyz)
XYZ
```

```python
XYZ.value_counts('XYZ')
```

```python
ABC = ABC.reset_index()
XYZ = XYZ.reset_index()
ABC
```

```python
matrix = ABC.merge(XYZ, on='StockCode')
matrix
```

```python
matrix['Grade'] = matrix['ABC_class'] + matrix['XYZ']
matrix
```

```python
matrix.value_counts('Grade')
```
